### Figure
- Overall window or page where everything is drawn on.
- It is the outermost container for all plots
- Can contain one or more **axes** which are the actual plots.

### Axes
- An area where data is plotted
- It contains most elements of a plot, such as the x-axis, y-axis, labels, ticks, and the plot itself.

### subplots
- It is a way of creating multiple plots on the same figure.
